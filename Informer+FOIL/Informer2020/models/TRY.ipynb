```python
import torch

# 假设的模型参数
batch_size = 2   # 批量大小
seq_len = 5      # 序列长度
feat_dim = 10    # 特征维度

# 模拟的从模型中获取的inv_emb_detach
inv_emb_detach = torch.randn(batch_size, seq_len, feat_dim)

# 现在我们假设你的模型有一个方法self.env_infer，它接收展平的inv_emb
# 我们模拟这个方法仅仅是将输入翻倍
class MockModel:
    def env_infer(self, x):
        return x * 2

# 初始化模拟模型
mock_model = MockModel()

# 展平 inv_emb_detach 以模拟通过模型的操作
inv_emb = inv_emb_detach.reshape(batch_size * seq_len, feat_dim)

# 通过假设的模型进行转换
env_w = mock_model.env_infer(inv_emb)

# 将结果重塑回原来的形状
env_w = env_w.reshape(batch_size, seq_len, -1)

# 打印出结果以验证
print("inv_emb_detach:", inv_emb_detach)
print("inv_emb (reshaped):", inv_emb)
print("env_w (after model inference):", env_w)
```
